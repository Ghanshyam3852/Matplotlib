```python
import matplotlib.pyplot as plt
```

# Create first Chart

```python
x = [1,2,3,4,5]
y = [10,15,20,50,80]
plt.plot(x,y)  #line chart draw
plt.show()        # Data show 
```

# Customize Chart

```python
plt.figure(figsize=(6,4))
plt.plot(x,y , color ='blue',marker ='o', linestyle ='dashed',linewidth = 2,markersize=16)
plt.title('Customize chart')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.show()
```

# Advanced line chart - multiple lines and legends

```python
x = [1,2,3,4,5]
y1 = [10,15,20,50,80]
y2 = [20,25,30,60,90]
plt.figure(figsize=(6,4))
plt.plot(x,y1, label='Sales 2025')
plt.plot(x,y2 , label='Sales 2026')
plt.title('yoy sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.legend()
plt.show()      
```

# Bar Chart 

```python
x = ['A','B','c','D','E']
y1 = [10,60,70,50,70]
y2 = [20,40,50,30,80]
plt.figure(figsize=(6,4))
plt.bar(x,y1 , label ='Sales 2024')#Bar chart
plt.bar(x,y2 , label='Sales 2025')
plt.title('Bar Chart')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.legend()
plt.show()        
```

# Histogram

```python
import random
data = [random.randint(1, 40) for _ in range(150)]

plt.hist(data , bins=5)
plt.title('Histogram')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.show()
```

# Pie Chart

```python
x = ['A','B','c','D','E']
y = [10,60,70,50,70]

plt.pie(y, labels = x, autopct ='%1.1f%%' , startangle =90)
plt.title('Pie chart')
plt.show()
```

# Scatter chart

```python
y1 = [10,20,25,35,45]
y2 = [20,30,35,45,55]

plt.scatter(y1,y2)
plt.title('Scatter Chart')
plt.show()
```

# Subplots

```python
x = ['Mon','Tue','Wed','Thu','Fri']
y = [10,60,70,50,70]

y1 = [10,20,25,35,45]
y2 = [20,30,35,45,55]
plt.figure(figsize=(10,4))

#first bar chart
plt.subplot(1,2,1)
plt.bar(x,y)
plt.title('Daily Sales')
plt.xlabel('Week Days')
plt.ylabel('Sales')

# Second Scatter Chart
plt.subplot(1,2,2)
plt.scatter(y1,y2)
plt.title('User Define')
plt.xlabel('User1')
plt.ylabel('User2')

plt.show()
```

# Matplotlib with Pandas - real data

```python
import pandas as pd
```

```python
data = {
    'Month' :['Jan','Feb','Mar','Apr'],
    'Sales' :[14000,34000,54000,34000]
}
df = pd. DataFrame(data)
df
```

```python
plt.bar(df['Month'],df['Sales'])
plt.title('Matplotlib with pandas')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

# Save Data

```python
plt.bar(df['Month'],df['Sales'])
plt.title('Matplotlib with pandas')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.savefig('Monthly_sales.png')
plt.show()
```
